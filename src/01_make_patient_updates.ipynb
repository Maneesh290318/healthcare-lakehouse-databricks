```python
import os
from pyspark.sql import functions as F

out_dir = "/Volumes/health/raw/synthea/batch_01_updates"
os.makedirs(out_dir, exist_ok=True)

src = spark.read.option("header", True).csv("/Volumes/health/raw/synthea/batch_01/patients.csv")
changed = (src.sample(fraction=0.05, seed=7)
           .withColumn("ADDRESS", F.concat(F.lit("100 Relocated Way Apt "), (F.rand(7) * 900).cast("int")))
           .withColumn("CITY", F.lit("Madison")))

changed.toPandas().to_csv(f"{out_dir}/patients.csv", index=False)
print(changed.count(), "patients moved to Madison")
```
